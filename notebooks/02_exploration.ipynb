## SCAN Framework - Exploratory Data Analysis

| SCAN Framework | California Property Close Price Prediction |
|---|---|
| **Stakeholder Goal** | Understand California single-family home sales and identify the major factors and data issues relevant to predicting `ClosePrice`. The eventual goal is to support the development of a reliable property price prediction model. |
| **Columns and Coverage** | Focus on `ClosePrice`, property characteristics such as `LivingArea`, bedrooms, bathrooms, lot size, and year built, along with geographic and time variables. Analyze at least 6 months of `Residential` → `SingleFamilyResidence` sales data. |
| **Aggregates and Anomalies** | Examine typical `ClosePrice` and property characteristics using counts, distributions, and summary statistics. Identify missing values, duplicates, impossible or suspicious values, extreme prices, and other potential outliers that could affect modeling. |
| **Notable Segments** | Compare `ClosePrice` across meaningful segments such as property characteristics, city/ZIP/MLS area, and month to identify patterns that may be useful for prediction. |

```python
# Check whether LotSizeSquareFeet = LotSizeAcres * 43,560
# Only rows where both lot-size fields are available
lot_check = df[['LotSizeAcres', 'LotSizeSquareFeet']].dropna().copy()

lot_check['ExpectedLotSqFt'] = lot_check['LotSizeAcres'] * 43560

np.isclose(
    lot_check['LotSizeSquareFeet'],
    lot_check['ExpectedLotSqFt']
).mean()
```

Lot-size data quality: Extremely large LotSizeSquareFeet values were identified. While several extreme observations are consistent with LotSizeAcres × 43,560, only approximately 14.9% of non-missing records satisfy this relationship overall. Because LotSizeUnits is not available in the provided dataset, the inconsistency cannot be fully resolved during EDA. Lot-size fields will therefore require further validation during preprocessing.

```python
#Number of sales, median close price, and median living area by month
monthly_summary = df.groupby('DataMonth').agg(
    number_of_sales=('ClosePrice', 'count'),
    median_close_price=('ClosePrice', 'median'),
    median_living_area=('LivingArea', 'median')
)

monthly_summary
```

```python
#Number of sales, mean close price, median close price, and median living area by city

city_summary = df.groupby('City').agg(
    number_of_sales=('ClosePrice', 'count'),
    mean_close_price=('ClosePrice', 'mean'),
    median_close_price=('ClosePrice', 'median'),
    median_living_area=('LivingArea', 'median')
).sort_values('number_of_sales', ascending=False)
city_summary
```

```python
#Number of sales, mean close price, median close price, and median living area by for cities with more than 100 sales
city_100_summary = df.groupby('City').agg(
    number_of_sales=('ClosePrice', 'count'),
    mean_close_price=('ClosePrice', 'mean'),
    median_close_price=('ClosePrice', 'median'),
    median_living_area=('LivingArea', 'median')
).sort_values('number_of_sales', ascending=False)

city_100_summary = city_100_summary[city_100_summary['number_of_sales'] > 100]
city_100_summary
```
